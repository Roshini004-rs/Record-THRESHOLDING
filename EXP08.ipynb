```python
#import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load Original Image
image = cv2.imread('pepper.jpg') 
```

```python
#Convert to GrayScale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
#Global thresholding
threshold_value = 127
ret, thresholded_image = cv2.threshold(gray_image,threshold_value,255,cv2.THRESH_BINARY)
```

```python
# Display the thresholded image
plt.figure(figsize=(10, 6))
plt.imshow(thresholded_image, cmap="gray")
plt.title("Global Thresholded Image")
plt.axis("off")
plt.show()
```

```python
#Adaptive Thresholding
# Adaptive Mean Thresholding
adaptive_mean = cv2.adaptiveThreshold(gray_image,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,11,2)

# Adaptive Gaussian Thresholding
adaptive_gaussian = cv2.adaptiveThreshold(gray_image,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
```

```python
# Display the segmented images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray_image, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(adaptive_mean, cmap="gray")
plt.title("Adaptive Mean Thresholding")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(adaptive_gaussian, cmap="gray")
plt.title("Adaptive Gaussian Thresholding")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 6: Otsu's Thresholding
otsu_threshold, otsu_image = cv2.threshold(gray_image,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Display the segmented image
plt.figure(figsize=(10, 6))
plt.imshow(otsu_image, cmap="gray")
plt.title(f"Otsu's Thresholding (Threshold = {otsu_threshold:.0f})")
plt.axis("off")
plt.show()

# Display the automatically calculated threshold
print("Optimal Otsu Threshold Value:", otsu_threshold)
```

```python
# Step 7: Compare Global, Adaptive, and Otsu's Thresholding
plt.figure(figsize=(15, 10))

# Global Thresholding
plt.subplot(2, 2, 1)
plt.imshow(thresholded_image, cmap="gray")
plt.title("Global Thresholding")
plt.axis("off")

# Adaptive Mean Thresholding
plt.subplot(2, 2, 2)
plt.imshow(adaptive_mean, cmap="gray")
plt.title("Adaptive Mean Thresholding")
plt.axis("off")

# Adaptive Gaussian Thresholding
plt.subplot(2, 2, 3)
plt.imshow(adaptive_gaussian, cmap="gray")
plt.title("Adaptive Gaussian Thresholding")
plt.axis("off")

# Otsu's Thresholding
plt.subplot(2, 2, 4)
plt.imshow(otsu_image, cmap="gray")
plt.title(f"Otsu's Thresholding (T = {otsu_threshold:.0f})")
plt.axis("off")

plt.tight_layout()
plt.show()
```
